```python
import numpy as np

temperature = np.array([30, 32, 29, 35, 33, 31, 28])

print("Temperature Data:", temperature)

print("Average Temperature:", np.mean(temperature))
print("Maximum Temperature:", np.max(temperature))
print("Minimum Temperature:", np.min(temperature))
print("Temperature Difference:", np.max(temperature) - np.min(temperature))

print("Days above Average:")
print(temperature[temperature > np.mean(temperature)])
```
