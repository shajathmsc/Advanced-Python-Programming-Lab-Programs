```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Cathy", "David"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nStudents with Marks above 80:")
print(df[df["Marks"] > 80])

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())
```
